## Objective

Create and validate Unity Catalog managed volumes for storing original source files before ingestion into Bronze Delta tables.

## Design principles

- Maintain separate raw storage for each official data provider.
- Preserve original files without modifying their contents.
- Use idempotent SQL statements to avoid duplicate volume creation.
- Keep raw landing storage separate from Bronze, Silver, and Gold Delta tables.
- Validate the existence of required volumes before beginning ingestion.

**Expected volumes:** `bcrd_raw`, `micm_raw`, `proconsumidor_raw`, `agriculture_raw`.

**Status:** Infrastructure setup; source ingestion not yet implemented.
