# 00 — Create Unity Catalog

**Project:** RD Economic & Financial Intelligence Platform  
**Layer:** Infrastructure / DDL  
**Environment:** Databricks Free Edition  
**Status:** Implemented and verified

## Objective
Create the Unity Catalog namespace used by the platform to organize Bronze, Silver, Gold, Metadata, and Quality resources.

## Design decisions
- Use a dedicated catalog named `rd_economic_intelligence`.
- Keep source files in managed Unity Catalog Volumes.
- Separate raw data, transformations, analytical outputs, and operational metadata.

## Validation
Verify the catalog is accessible and list its schemas.

## Execution notes
This setup uses idempotent DDL statements to avoid recreating existing objects.

```python
%sql

CREATE CATALOG IF NOT EXISTS rd_economic_intelligence;
  
```
