```python

from pathlib import Path

base = Path("/Volumes/rd_economic_intelligence/bronze")

datasets = {
    "bcrd_raw": [
        "exchange_rates",
        "cpi",
        "monetary_policy"
    ],
    "micm_raw": [
        "fuel_prices"
    ],
    "proconsumidor_raw": [
        "consumer_prices"
    ],
    "agriculture_raw": [
        "retail_prices",
        "wholesale_prices"
    ]
}

for volume, folders in datasets.items():
    for folder_name in folders:
        path = base / volume / folder_name
        path.mkdir(parents=True, exist_ok=True)
        print(f"OK: {path}")
```
