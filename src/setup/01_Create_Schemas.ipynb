# 01 — Unity Catalog Schema Setup

**Project:** RD Economic & Financial Intelligence Platform  
**Environment:** Databricks Free Edition  
**Component:** Infrastructure / DDL  
**Catalog:** `rd_economic_intelligence`  
**Status:** Implemented and verified

## Objective

Define and validate the five logical schemas supporting the Medallion architecture, operational metadata, and data quality management.

## Schema responsibilities

| Schema | Purpose |
|---|---|
| Bronze | Source-aligned Delta tables and ingestion metadata |
| Silver | Cleaned, standardized, and validated datasets |
| Gold | Business-ready analytical models and KPIs |
| Metadata | Source registry, ingestion runs, file manifests, and watermarks |
| Quality | Data quality results, exceptions, and quarantined records |

## Design considerations

- Use `IF NOT EXISTS` to make schema creation rerunnable.
- Preserve Databricks-managed `information_schema` and existing `default`.
- Keep Unity Catalog Volumes separate from Delta tables.
- Do not create source tables until their schemas and grain are documented.

## Validation

Run `SHOW SCHEMAS IN rd_economic_intelligence` and confirm all five project schemas exist.

## Dependencies

Requires the `rd_economic_intelligence` catalog.

```python

catalog = "rd_economic_intelligence"

schemas = ["bronze", "silver", "gold", "metadata", "quality"]

for schema in schemas:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")
    print(f"Schema ready: {catalog}.{schema}")
```

```python

existing_schemas = {
    row.databaseName
    for row in spark.sql(
        "SHOW SCHEMAS IN rd_economic_intelligence"
    ).collect()
}

required_schemas = {"bronze", "silver", "gold", "metadata", "quality"}
missing = required_schemas - existing_schemas

display(spark.sql("SHOW SCHEMAS IN rd_economic_intelligence"))

assert not missing, f"Missing schemas: {sorted(missing)}"
print("PASS: All five project schemas are available.")
```
